# SYB Network Experience

Welcome to the SYB (Sybil Resistance) Network! This interactive notebook lets you experience how the SYB algorithm works in a network with a dynamic, real-world simulation.

## What is SYB?
SYB is a decentralized protocol that uses social vouching and algorithmic scoring to prevent sybil attacks. Users deposit funds, vouch for others they trust, and the network calculates reputation scores using specific algorithms devised to attain sybil resistance.

### Key uses
There are several use cases for such a network environment, but the two main ones are:
    - proof of identity which is not based on centralized-issued documents or biometric information;
    - voting system, reputation scores for economic and social purposes.

## Key Features
The aim of this notebook is to be as close as possible to the real-world implementation.
You, the user, are `user 0`, and you are equipped with:
- **Balance**: the amount of TokamakTON you are storing in the contract; the protocol requires that a minimum Balance threshold is satisfied in order to interact with the network.
- **Score**: the actual reputation score (see more info [here](no_link)).

You, the user, start as a node in the network, with a Balance of 0 TokamakTON. You can then interact with the network by the following actions:
- **Deposit**: deposits a selected amount into the node balance (note that we do not provide any upper bound on this quantity).
- **Withdraw**: withdraw a certain amount of TokamakTON from the balance present in the node; note that if the node is part of the network, your current balance must be at least 1 Tokamak WTON. If the node is disjoint from the network, all the balance can be withdrawn.
- **Vouch**: vouch for a node; note that your current balance must be at least 1 Tokamak WTON in order to be able to vouch for any other node.
- **Unvouch**: unvouch from a previously vouched for node; note that your current balance must be at least 1 Tokamak WTON in order to be able to vouch for any other node.

In the real-world implementation of the SYB protocol, we process trnasactions in batches of at least 5 transactions.


### Extra features
